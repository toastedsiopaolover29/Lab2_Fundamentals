```python
# [CEll 1] CONFIGURATION
SURNAME = "Bautista"
STUDENT_ID = "TUPM-26-3506"

# Derivation: Extract the last digit to acTt as my Logic Seed
SEED_NUM = int(STUDENT_ID[-1])  # Extracts the last digit of the STUDENT_ID

print(f"Acquired Surname: {SURNAME}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [Cell 2] SENSOR DATA GENERATION
# Convert surname letters to ASCII values
ascii_values = [ord(ch) for ch in SURNAME]

# Transform using SEED_NUM
readings = [(val * SEED_NUM) % 150 for val in ascii_values]  # bounded within 0–149

# Format to fixed length (3 digits)
fixed_readings = [str(r).zfill(3) for r in readings]

print(f"Transformed Values: {readings}")
```

```python
# [Cell 3] CLASSIFICATION

results = []  # Create an empty list to store each reading and its classification

for r in readings:  # Go through each sensor reading one by one
    # Check if the reading values are between 0 and 100
    # If yes, classify it as "Valid"; otherwise, classify it as "Invalid"
    status = "Valid" if 0 <= r <= 100 else "Invalid"
    
    # Store the sensor reading and its classification as a tuple
    results.append((r, status))
```

```python
# [Cell 4] SUMMARY

# Display the sensor readings in their fixed 3-digit format
print(f"Sensor Readings (Fixed Length): {fixed_readings}")

# Display each sensor reading together with its classification
print(f"Classification Results: {results}")

# Count how many readings were classified as "Valid"
valid_count = sum(1 for _, status in results if status == "Valid")

# Calculate the number of Invalid readings
# by subtracting the number of Valid readings from the total number of readings
invalid_count = len(results) - valid_count

# Display the total number of Valid and Invalid readings
print(f"Valid: {valid_count}, Invalid: {invalid_count}")
```
