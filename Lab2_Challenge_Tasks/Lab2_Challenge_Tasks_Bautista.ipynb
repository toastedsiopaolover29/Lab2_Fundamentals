```python
# [CEll 1] CONFIGURATION
SURNAME = "Bautista"
STUDENT_ID = "TUPM-26-3506"

# Derivation: Extract the last digit to acTt as my Logic Seed
SEED_NUM = int(STUDENT_ID[-1])  # Extracts the last digit of the STUDENT_ID

print(f"Acquired Surname: {SURNAME}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] SIGNAL GENERATION & NORMALIZATION
# first + last letter of surname + seed
signal = SURNAME[0] + SURNAME[-1] + str(SEED_NUM)   # "Ba6"
processed_signal = signal.strip().upper()
```

```python
# [CELL 3] CHARACTER ANALYSIS
analysis = []
for ch in processed_signal:
    if ch.isalpha():
        ctype = "Letter"
    elif ch.isdigit():
        ctype = "Digit"
    else:
        ctype = "Special"
    analysis.append((ch, ctype, ord(ch)))
```

```python
# [CELL 4] DISPLAY
letters = sum(1 for _, ctype, _ in analysis if ctype == "Letter")
digits = sum(1 for _, ctype, _ in analysis if ctype == "Digit")

print("Generated Signal:", signal)
print("Processed Signal:", processed_signal)
print("Character Analysis:", analysis)
print(f"Final Output: Signal '{processed_signal}' contains {letters} letters and {digits} digit(s).")
```
