```python
# [CEll 1] CONFIGURATION
SURNAME = "Bautista"
STUDENT_ID = "TUPM-26-3506"

# Derivation: Extract the last digit to acTt as my Logic Seed
SEED_NUM = int(STUDENT_ID[-1])  # Extracts the last digit of the STUDENT_ID

print(f"Acquired Surname: {SURNAME}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] PASSWORD AND LIMIT GENERATION
first_three = SURNAME[:3].upper()
ascii_plus_one = ord(SURNAME[0].upper()) + 1
password = first_three + str(SEED_NUM) + str(ascii_plus_one)

# Attempt limit
attempt_limit = SEED_NUM // 2
```

```python
# [CELL 3] AUTHENTICATION WORKFLOW
attempts = 0
access_granted = False

while attempts < attempt_limit:
    user_input = input("Enter password: ")
    attempts += 1
    if user_input == password:
        access_granted = True
        break
```

```python
# [Cell 4] FINAL STATE
if access_granted:
    system_state = "Unlocked"
    result = "Access Granted"
else:
    system_state = "Locked"
    result = "Access Denied"
```

```python
# [Cell 5] DISPLAY
print("Generated Password:", password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts)
print("Access Result:", result)
print("Final System State:", system_state)
print("Final Output:", f"System ended in state: {system_state}")
```
